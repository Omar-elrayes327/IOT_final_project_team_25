```python
import numpy as np
import pandas as pd
import random
```

```python
good_news_df = {"Demand": [40, 50, 60, 70,80,90,100], "prob": [0.03, 0.05, 0.15, 0.2,0.35,0.15,0.07], "ranges": [range(1, 4), range(4, 9), range(9, 24), range(24, 44),range(44,79),range(79,94),range(94,101)]}
poor_news_df = {"Demand": [40, 50, 60, 70, 80, 0, 0], "prob": [0.44, 0.22, 0.16, 0.12,0.06, 0.0, 0.0], "ranges": [range(1, 45), range(45, 67), range(67, 83), range(83, 95),range(95,101),[0],[0]]}
fair_news_df = {"Demand": [40, 50, 60, 70, 80, 90, 0], "prob": [0.1, 0.18, 0.40, 0.20, 0.08, 0.04, 0.0], "ranges": [range(1, 11), range(11, 29), range(29, 69),range(69,89),range(89,97),range(97,101), [0]]}
news_type_df = {"News_Type": ["Good", "Fair", "Poor"], "prob": [0.35, 0.45, 0.2], "ranges": [range(1, 36), range(36, 81), range(81, 101)]}
names_df={"Good":good_news_df, "Fair":fair_news_df, "Poor":poor_news_df}
```

```python
news_simulation_df = {"Day":[i for i in range(1,51)],"RN_(NT)":[],"News_Type":[],"RN_(Demand)":[],"Demand":[],"news_cost_(per one)_(in_cents)":[],"revenue_(in_cents)":[],"lost_profit_(in_cents)":[],"scrap_rev_(in_cents)":[],"Daily profit_(in_Dollar)":[]}

for day_idx in range(50):

    rn_nt = random.randint(1,100)
    rn_demand = random.randint(1,100)

    news_simulation_df["RN_(NT)"].append(rn_nt)
    news_simulation_df["RN_(Demand)"].append(rn_demand)

    current_news_type = ""
    for j, r_range in enumerate(news_type_df["ranges"]):
        if rn_nt in r_range:
            current_news_type = news_type_df["News_Type"][j]
            break
    news_simulation_df["News_Type"].append(current_news_type)

    news_simulation_df["news_cost_(per one)_(in_cents)"].append(34) # This value (34) is acting as selling price/profit per unit

    df_iter=names_df[current_news_type]
    demand_value = 0
    for k in df_iter["ranges"]:
      if rn_demand in k:
        demand_value = df_iter["Demand"][df_iter["ranges"].index(k)]
        break


    news_simulation_df["Demand"].append(demand_value)
    revenue_value = news_simulation_df["news_cost_(per one)_(in_cents)"][day_idx] * demand_value
    news_simulation_df["revenue_(in_cents)"].append(revenue_value)
    remain=0
    remain=70-news_simulation_df["Demand"][day_idx]
    if remain >= 0:
        current_scrap_rev = remain * 35
        news_simulation_df["scrap_rev_(in_cents)"].append(current_scrap_rev)
        news_simulation_df["lost_profit_(in_cents)"].append(0)
    else  :
        lost = remain * 34
        news_simulation_df["lost_profit_(in_cents)"].append(lost)
        news_simulation_df["scrap_rev_(in_cents)"].append(0)



    # print(f"Day {day_idx}: len(revenue)={len(news_simulation_df['revenue'])}, len(lost_profit)={len(news_simulation_df['lost_profit'])}, len(scrap_rev)={len(news_simulation_df['scrap_rev'])}")
    daily_prof = ((news_simulation_df["revenue_(in_cents)"][day_idx] +
                  news_simulation_df["lost_profit_(in_cents)"][day_idx] +
                  news_simulation_df["scrap_rev_(in_cents)"][day_idx]) - news_simulation_df["news_cost_(per one)_(in_cents)"][day_idx])/100
    news_simulation_df["Daily profit_(in_Dollar)"].append(daily_prof)


news_simulation_df = pd.DataFrame(news_simulation_df)
```
